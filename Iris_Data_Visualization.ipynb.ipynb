```python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm",
    "Species"
]

df = pd.read_csv(url, names=columns)

df.insert(0, "Id", range(1, len(df) + 1))

df.to_csv("Iris.csv", index=False)

df.head()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDataset Information:")
df.info()
```

```python
avg_petal_length = df.groupby("Species")["PetalLengthCm"].mean()

plt.figure(figsize=(8, 5))
avg_petal_length.plot(kind="bar")

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("average_petal_length.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["PetalLengthCm"], bins=10)

plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig("petal_length_histogram.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["SepalLengthCm"], bins=10)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig("sepal_length_histogram.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

for species in df["Species"].unique():
    data = df[df["Species"] == species]
    plt.scatter(
        data["PetalLengthCm"],
        data["PetalWidthCm"],
        label=species
    )

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend()

plt.tight_layout()
plt.savefig("petal_length_vs_width.png")
plt.show()
```

🔎 Key Insights
1. Iris-setosa has the smallest average petal length.
2. Iris-virginica has the largest average petal length.
3. Petal length and petal width show a clear relationship across the species.
4. Sepal measurements have more overlap between the species.
5. Data visualization helps identify patterns and relationships easily.
